### Codificamos..

Aplicamos el tipo de dato `category` de pandas a las variables nominales (incluyendo las de alta cardinalidad que luego usará la capa de embeddings), para un manejo más eficiente de memoria.

```python
df.columns = df.columns.str.strip()
nominal_cols = ['workclass', 'marital-status', 'occupation', 'relationship',
                'race', 'native-country', 'skill-profile']

df[nominal_cols] = df[nominal_cols].astype('category')
```

Convertimos `education` en una categoría ordinal, usando el orden definido y validado más arriba.

```python
education_order = pd.CategoricalDtype(categories=education_categories, ordered=True)
df['education'] = df['education'].astype(education_order).cat.codes
```

Aplicamos one-hot encoding a las columnas de baja cardinalidad (<10 valores únicos, sin orden lógico).

```python
low_card_cols = ['race', 'relationship', 'workclass', 'marital-status']
df = pd.get_dummies(df, columns=low_card_cols, drop_first=True)
```

```python
df['sex'] = (df['sex'].str.strip() == 'Male').astype(int)
df['sex'].value_counts()
```

```python
df['income'] = (df['income'].str.strip() == '>50K').astype(int)
df['income'].value_counts()
```

`occupation`, `native-country` y `skill-profile` quedan con dtype `category` (aplicado más arriba) y se codifican mediante una capa de `nn.Embedding`.

Aplicamos las mismas transformaciones en adult_val.csv

```python
val_df.columns = val_df.columns.str.strip()

val_df[nominal_cols] = val_df[nominal_cols].astype('category')
# reusamos las categorías de train para que los codes de occupation/native-country/skill-profile
# queden alineados con los índices que ya usan los embeddings entrenados
for col in ['occupation', 'native-country', 'skill-profile']:
    val_df[col] = val_df[col].cat.set_categories(df[col].cat.categories)

val_df['education'] = val_df['education'].astype(education_order).cat.codes

val_df = pd.get_dummies(val_df, columns=low_card_cols, drop_first=True)

val_df['sex'] = (val_df['sex'].str.strip() == 'Male').astype(int)
val_df['income'] = (val_df['income'].str.strip() == '>50K').astype(int)

# alineamos columnas y orden con df por si alguna dummy no aparece en val
val_df = val_df.reindex(columns=df.columns, fill_value=0)
```

### Definimos el modelo

```python
# usamos un seed fijo para reproducibilidad
seed = 24
random.seed(seed); np.random.seed(seed); torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)

class PredictIncomeModel(nn.Module):
    def __init__(self, input_dim, p=0.2):
        super(PredictIncomeModel, self).__init__()
        # seteamos embedding_dim en 4, 8, 8 inicialmente, pero se puede ajustar según la experimentación
        self.occupation_embedding = nn.Embedding(num_embeddings=df['occupation'].nunique(), embedding_dim=4)
        self.native_country_embedding = nn.Embedding(num_embeddings=df['native-country'].nunique(), embedding_dim=8)
        self.skill_profile_embedding = nn.Embedding(num_embeddings=df['skill-profile'].nunique(), embedding_dim=8)

        self.fc = nn.Sequential(
            nn.Linear(input_dim + 4 + 8 + 8, 64),
            nn.ReLU(),
            nn.Dropout(p),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(p),
            nn.Linear(32, 1),
            nn.Sigmoid()
        )

    def forward(self, x, occupation, native_country, skill_profile):
        occupation_embedded = self.occupation_embedding(occupation)
        native_country_embedded = self.native_country_embedding(native_country)
        skill_profile_embedded = self.skill_profile_embedding(skill_profile)

        x = torch.cat([x, occupation_embedded, native_country_embedded, skill_profile_embedded], dim=1)
        return self.fc(x)
```

Definimos optimizer y loss function e instanciamos el modelo

```python
import torch.optim as optim
import torch.nn as nn
import torch
import math
from sklearn.metrics import accuracy_score, f1_score

feature_cols = df.drop(columns=['income', 'occupation', 'native-country', 'skill-profile']).columns
input_dim = len(feature_cols)

model = PredictIncomeModel(input_dim=input_dim)
optimizer = optim.Adam(model.parameters(), lr=0.001)
criterion = nn.BCELoss()  # Binary CrossEntropyLoss

def rmse(y_true, y_pred):
    return math.sqrt(((y_true - y_pred)**2).mean().item())
```

```python
epochs = 30
batch_size = 64

train_mse = []
val_mse = []
train_acc, val_acc = [], []
train_f1, val_f1 = [], []

for epoch in range(1, epochs+1):
    model.train()
    train_preds, train_true = [], []
    train_loss = 0.0
    # minibatch training loop en batches de 64, ~471 updates por epoca
    for i in range(0, len(df), batch_size):
        batch = df.iloc[i:i+batch_size]
        x_batch = torch.tensor(batch.drop(columns=['income', 'occupation', 'native-country', 'skill-profile']).to_numpy(dtype=np.float32))
        occupation_batch = torch.tensor(batch['occupation'].cat.codes.values, dtype=torch.long) # extrae los codigos numéricos de occupation que fue casteadad más arriba
        native_country_batch = torch.tensor(batch['native-country'].cat.codes.values, dtype=torch.long) # extrae los codigos numéricos de native-country que fue casteadad más arriba
        skill_profile_batch = torch.tensor(batch['skill-profile'].cat.codes.values, dtype=torch.long) # extrae los codigos numéricos de skill-profile que fue casteadad más arriba
        y_batch = torch.tensor(batch['income'].values, dtype=torch.float32).unsqueeze(1)

        optimizer.zero_grad()
        outputs = model(x_batch, occupation_batch, native_country_batch, skill_profile_batch)
        loss = criterion(outputs, y_batch)
        loss.backward()
        optimizer.step()

        train_loss += loss.item() * x_batch.size(0)
        preds = (outputs >= 0.5).float() # convierte las probabilidades a binario: 0.0 o 1.0
        train_preds.extend(preds.detach().numpy().flatten())
        train_true.extend(y_batch.numpy().flatten())

    train_loss /= len(df)
    train_mse.append(train_loss)
    t_acc = accuracy_score(train_true, train_preds)
    t_f1 = f1_score(train_true, train_preds, average='macro')
    train_acc.append(t_acc)
    train_f1.append(t_f1)

    # Validation
    model.eval()
    val_preds, val_true = [], []
    val_loss = 0.0
    with torch.no_grad():
        for i in range(0, len(val_df), batch_size):
            batch = val_df.iloc[i:i+batch_size]
            x_batch = torch.tensor(batch.drop(columns=['income', 'occupation', 'native-country', 'skill-profile']).to_numpy(dtype=np.float32))
            occupation_batch = torch.tensor(batch['occupation'].cat.codes.values, dtype=torch.long)
            native_country_batch = torch.tensor(batch['native-country'].cat.codes.values, dtype=torch.long)
            skill_profile_batch = torch.tensor(batch['skill-profile'].cat.codes.values, dtype=torch.long)
            y_batch = torch.tensor(batch['income'].values, dtype=torch.float32).unsqueeze(1)

            outputs = model(x_batch, occupation_batch, native_country_batch, skill_profile_batch)
            loss = criterion(outputs, y_batch)

            val_loss += loss.item() * x_batch.size(0)
            preds = (outputs >= 0.5).float()
            val_preds.extend(preds.detach().numpy().flatten())
            val_true.extend(y_batch.numpy().flatten())

    val_loss /= len(val_df)
    val_mse.append(val_loss)

    v_acc = accuracy_score(val_true, val_preds)
    v_f1 = f1_score(val_true, val_preds, average='macro')
    val_acc.append(v_acc)
    val_f1.append(v_f1)

    print(f'Epoch {epoch}/{epochs}, Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}')
```

## c) Diseño y entrenamiento de un modelo sin embeddings (3 puntos)

- Entrenar un segundo modelo, aplicando one-hot encoding a todas las variables que en el punto b) fueron representadas mediante embeddings.
- Quitar todas las capas de dropout, para este segundo modelo no se usará ninguna capa interna de regularización.
- Con respecto al resto de capas, se debe mantener exactamente la misma arquitectura del modelo anterior: igual número de capas, misma cantidad de neuronas por capa y mismas funciones de activación.
- Presentar las mismas métricas, visualizaciones y reportes que en el modelo con embeddings.

Preparación de los datos con One-Hot Encoding

```python
# Aplicamos One-Hot Encoding a las variables categóricas
cat_cols = ['occupation', 'native-country', 'skill-profile']
df_all = pd.concat([df, val_df], axis=0)
df_all_encoded = pd.get_dummies(df_all, columns=cat_cols, drop_first=False)

train_df_oh = df_all_encoded.iloc[:len(df)].copy()
val_df_oh = df_all_encoded.iloc[len(df):].copy()

input_dim_oh = train_df_oh.drop(columns=['income']).shape[1]
```

Definición de Arquitectura

```python
# Definición del modelo sin embeddings y sin dropout (misma arquitectura)
input_dim_oh = train_df_oh.drop(columns=['income']).shape[1]

model_no_emb = nn.Sequential(
    nn.Linear(input_dim_oh, 256),
    nn.LeakyReLU(),
    nn.Linear(256, 128),
    nn.LeakyReLU(),
    nn.Linear(128, 64),
    nn.LeakyReLU(),
    nn.Linear(64, 1),
    nn.Sigmoid()
)

optimizer_no_emb = optim.Adam(model_no_emb.parameters(), lr=0.001)
criterion_no_emb = nn.BCELoss()
```

Entrenamiento y Evaluación

```python
epochs = 30
batch_size = 64

train_mse_no_emb, val_mse_no_emb = [], []
train_acc_no_emb, val_acc_no_emb = [], []
train_f1_no_emb, val_f1_no_emb = [], []

for epoch in range(1, epochs+1):
    model_no_emb.train()
    train_loss = 0.0
    train_preds, train_true = [], []

    for i in range(0, len(train_df_oh), batch_size):
        batch = train_df_oh.iloc[i:i+batch_size]
        x_batch = torch.tensor(batch.drop(columns=['income']).to_numpy(dtype=np.float32))
        y_batch = torch.tensor(batch['income'].values, dtype=torch.float32).unsqueeze(1)

        optimizer_no_emb.zero_grad()
        outputs = model_no_emb(x_batch)
        loss = criterion_no_emb(outputs, y_batch)
        loss.backward()
        optimizer_no_emb.step()

        train_loss += loss.item() * x_batch.size(0)

        preds = (outputs >= 0.5).float()
        train_preds.extend(preds.detach().numpy().flatten())
        train_true.extend(y_batch.numpy().flatten())

    train_loss /= len(train_df_oh)
    train_mse_no_emb.append(train_loss)

    t_acc = accuracy_score(train_true, train_preds)
    t_f1 = f1_score(train_true, train_preds, average='macro')
    train_acc_no_emb.append(t_acc)
    train_f1_no_emb.append(t_f1)

    # Validación
    model_no_emb.eval()
    val_loss = 0.0
    val_preds, val_true = [], []

    with torch.no_grad():
        for i in range(0, len(val_df_oh), batch_size):
            batch = val_df_oh.iloc[i:i+batch_size]
            x_batch = torch.tensor(batch.drop(columns=['income']).to_numpy(dtype=np.float32))
            y_batch = torch.tensor(batch['income'].values, dtype=torch.float32).unsqueeze(1)

            outputs = model_no_emb(x_batch)
            loss = criterion_no_emb(outputs, y_batch)

            val_loss += loss.item() * x_batch.size(0)

            preds = (outputs >= 0.5).float()
            val_preds.extend(preds.detach().numpy().flatten())
            val_true.extend(y_batch.numpy().flatten())

    val_loss /= len(val_df_oh)
    val_mse_no_emb.append(val_loss)

    v_acc = accuracy_score(val_true, val_preds)
    v_f1 = f1_score(val_true, val_preds, average='macro')
    val_acc_no_emb.append(v_acc)
    val_f1_no_emb.append(v_f1)

    print(f'Epoch {epoch}/{epochs} | Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}')
```

Gráficas y Matrices de Confusión

```python
# 1. Curvas de Pérdida (Loss)
plt.figure(figsize=(7,4))
plt.plot(range(1, epochs+1), train_mse_no_emb, marker="o", label="Train Loss")
plt.plot(range(1, epochs+1), val_mse_no_emb, marker="o", label="Val Loss")
plt.xlabel("Época")
plt.ylabel("Pérdida")
plt.title("Evolución de Loss (Sin Embeddings / Sin Dropout)")
plt.grid(True, linestyle=":")
plt.legend()
plt.tight_layout()
plt.show()

# 2. Curvas de Accuracy
plt.figure(figsize=(7,4))
plt.plot(range(1, epochs+1), train_acc_no_emb, marker="o", label="Train Accuracy")
plt.plot(range(1, epochs+1), val_acc_no_emb, marker="o", label="Val Accuracy")
plt.xlabel("Época")
plt.ylabel("Accuracy")
plt.title("Evolución del Accuracy (Sin Embeddings / Sin Dropout)")
plt.grid(True, linestyle=":")
plt.legend()
plt.tight_layout()
plt.show()

# 3. Curvas de F1 Macro
plt.figure(figsize=(7,4))
plt.plot(range(1, epochs+1), train_f1_no_emb, marker="o", label="Train F1 Macro")
plt.plot(range(1, epochs+1), val_f1_no_emb, marker="o", label="Val F1 Macro")
plt.xlabel("Época")
plt.ylabel("F1 Macro")
plt.title("Evolución del F1 Macro (Sin Embeddings / Sin Dropout)")
plt.grid(True, linestyle=":")
plt.legend()
plt.tight_layout()
plt.show()

# 4. Evaluación final
model_no_emb.eval()
preds_list_no_emb = []
true_list_no_emb = []

with torch.no_grad():
    for i in range(0, len(val_df_oh), batch_size):
        batch = val_df_oh.iloc[i:i+batch_size]
        x_b = torch.tensor(batch.drop(columns=['income']).to_numpy(dtype=np.float32))
        y_b = torch.tensor(batch['income'].values, dtype=torch.float32).unsqueeze(1)

        out = model_no_emb(x_b)
        preds = (out >= 0.5).float()

        preds_list_no_emb.extend(preds.numpy().flatten())
        true_list_no_emb.extend(y_b.numpy().flatten())

# Classification Report
print(classification_report(true_list_no_emb, preds_list_no_emb, target_names=['<=50K', '>50K']))

# 5. Matrices de Confusión
cm = confusion_matrix(true_list_no_emb, preds_list_no_emb)
cm_norm = confusion_matrix(true_list_no_emb, preds_list_no_emb, normalize='true')

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0],
            xticklabels=['<=50K', '>50K'], yticklabels=['<=50K', '>50K'])
axes[0].set_title('Matriz de Confusión Absoluta (Sin Embeddings)')
axes[0].set_xlabel('Predicción')
axes[0].set_ylabel('Real')

sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='Blues', ax=axes[1],
            xticklabels=['<=50K', '>50K'], yticklabels=['<=50K', '>50K'])
axes[1].set_title('Matriz de Confusión Normalizada (Sin Embeddings)')
axes[1].set_xlabel('Predicción')
axes[1].set_ylabel('Real')

plt.tight_layout()
plt.show()
```
